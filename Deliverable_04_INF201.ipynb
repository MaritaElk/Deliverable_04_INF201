# Task 1: Temperatures

This task implements a `Temperature` class in Python that supports Celsius (C), Fahrenheit (F), and Kelvin (K).

The class provides temperature conversions, input validation, read-only attributes, string representations, and comparison operators.

The implementation is located in `temperature.py`, while the automated tests are implemented in `test_temperature.py` using pytest.

The following sections demonstrate the functionality of the class and present the results of the automated tests.

## Class Implementation

The `Temperature` class is implemented in `temperature.py`.

The class is initialized with a numeric value and a temperature unit (`C`, `F`, or `K`), with Celsius as the default unit.

The implementation includes:

- **Input validation:** Invalid units and temperatures below absolute zero raise a `ValueError`.
- **Read-only attributes:** The temperature value and unit are accessible through `@property`.
- **Temperature conversions:** Methods convert temperatures between Celsius, Fahrenheit, and Kelvin.
- **Object conversion:** `convert_to()` returns a new `Temperature` instance without modifying the original.
- **String representations:** `__repr__()` provides an unambiguous representation, while `__str__()` provides a readable format with two decimal places.
- **Comparison operators:** Temperatures can be compared across different units using a floating-point tolerance of `1e-9`.

The `@total_ordering` decorator from `functools` is used to generate the comparison operators from `__eq__()` and `__lt__()`.

```python
from temperature import Temperature

temp = Temperature(25, "C")

print("Temperature object:", temp)
print("Value:", temp.value)
print("Unit:", temp.unit)
print("Representation:", repr(temp))
```

## Temperature Conversions

The `Temperature` class provides three conversion methods:

- `to_celsius()`
- `to_fahrenheit()`
- `to_kelvin()`

The following example demonstrates the conversions using the five reference temperatures specified in the assignment.

```python
reference_temperatures = {
    "Absolute zero": -273.15,
    "Celsius-Fahrenheit crossover": -40.0,
    "Freezing point of water": 0.0,
    "Human body temperature": 37.0,
    "Boiling point of water": 100.0,
}

print("Temperature Conversions\n")
print(f"{'Reference Point':<32} {'Celsius':>12} {'Fahrenheit':>14} {'Kelvin':>12}")
print("-" * 74)

for name, value in reference_temperatures.items():
    temp = Temperature(value, "C")

    print(
        f"{name:<32} "
        f"{temp.to_celsius():>12.2f} "
        f"{temp.to_fahrenheit():>14.2f} "
        f"{temp.to_kelvin():>12.2f}"
    )
```

## Validation and Read-Only Attributes

The `Temperature` class validates input values during initialization.

- Invalid temperature units raise a `ValueError`.
- Temperatures below absolute zero raise a `ValueError`.
- The `value` and `unit` attributes are read-only and cannot be reassigned directly.

The following examples demonstrate these restrictions.

```python
print("=== Input Validation ===")

invalid_temperatures = [
    (25, "X"),
    (-274, "C"),
    (-460, "F"),
    (-1, "K"),
]

for value, unit in invalid_temperatures:
    try:
        Temperature(value, unit)
    except ValueError as error:
        print(f"{value} {unit}: {error}")


print("\n=== Read-Only Attributes ===")

temp = Temperature(25, "C")

try:
    temp.value = 100
except AttributeError:
    print("Value cannot be modified: PASSED")

try:
    temp.unit = "F"
except AttributeError:
    print("Unit cannot be modified: PASSED")

print(f"\nOriginal temperature: {temp}")
```

## Comparison Operators

The `Temperature` class supports all six comparison operators (`==`, `!=`, `<`, `<=`, `>`, `>=`).

Temperatures are converted to Kelvin before comparison, allowing objects with different units to be compared directly.

The comparisons use a tolerance of `1e-9` to account for floating-point precision.

The `@total_ordering` decorator generates the remaining ordering operators from `__eq__()` and `__lt__()`.

```python
print("=== Cross-Unit Comparisons ===")

cold = Temperature(0, "C")
equal = Temperature(32, "F")
warm = Temperature(300, "K")

print("0 °C == 32 °F:", cold == equal)
print("0 °C != 300 K:", cold != warm)
print("0 °C < 300 K:", cold < warm)
print("0 °C <= 32 °F:", cold <= equal)
print("300 K > 0 °C:", warm > cold)
print("32 °F >= 0 °C:", equal >= cold)

print("\n=== Floating-Point Tolerance ===")

temp1 = Temperature(20.0, "C")
temp2 = Temperature(20.0 + 1e-10, "C")

print("Equal within tolerance:", temp1 == temp2)
print("Not less within tolerance:", not (temp1 < temp2))
print("Not greater within tolerance:", not (temp1 > temp2))
```
