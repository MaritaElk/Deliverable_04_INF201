**Class Implementation**

The `Fraction` class is implemented in `my_fraction.py`.

The class is initialized with an integer numerator and an integer denominator. The denominator must be non-zero and is always normalized to be strictly positive. Fractions are automatically simplified during initialization.

The implementation includes:

- **Input validation:** Non-integer values, including booleans, raise a `TypeError`. A denominator of zero raises a `ValueError`.
- **Fraction simplification:** Fractions are automatically reduced to their simplest form using `gcd`.
- **String representations:** `__repr__()` provides an unambiguous representation, while `__str__()` provides a readable fraction format.
- **Comparison operators:** All six comparison operators are implemented: equal to, not equal to, less than, less than or equal to, greater than, and greater than or equal to.
- **Arithmetic operations:** Four special methods support addition (`__add__()`), subtraction (`__sub__()`), multiplication (`__mul__()`), and division (`__truediv__()`).
- **Error handling:** Division by zero raises a `ZeroDivisionError`.

The automated tests are implemented in `test_my_fraction.py` using pytest.

**Example Usage**

The following example demonstrates how to create fractions, perform arithmetic operations, and compare two fractions.

```python
from my_fraction import Fraction

# Create two fractions
fraction1 = Fraction(1, 2)
fraction2 = Fraction(3, 4)

# Display the fractions
print(fraction1)
print(fraction2)

# Perform arithmetic operations
print(fraction1 + fraction2)  # Addition
print(fraction1 - fraction2)  # Subtraction
print(fraction1 * fraction2)  # Multiplication
print(fraction1 / fraction2)  # Division

# Compare fractions
print(fraction1 < fraction2)
print(fraction1 == fraction2)
```
